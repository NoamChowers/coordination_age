```python
from collections import defaultdict


def infer_experiment(description):
    description = str(description).lower()

    if 'box and blocks' in description:
        return 'Box and Blocks'
    if 'jebsen-taylor' in description:
        return 'Jebsen-Taylor'
    if 'finger pressing' in description:
        return 'Finger pressing'
    if 'reach-to-grasp' in description:
        return 'Reach-to-grasp'
    if 'lifting' in description:
        return 'Lifting'

    return None


variable_col = variables_df.columns[0]
description_col = variables_df.columns[1]

experiment_to_columns = defaultdict(list)

for column, description in variables_df.iloc[3:][[variable_col, description_col]].itertuples(index=False, name=None):
    experiment = infer_experiment(description)
    experiment_to_columns[experiment].append(column)


experiment_to_columns = dict(experiment_to_columns)
experiment_to_columns
```

```python
import pandas as pd
import numpy as np
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.pipeline import make_pipeline

def pca_reduce_groups(df: pd.DataFrame, groups: dict) -> pd.DataFrame:
    """For each group of columns, impute (mean), standardize, run PCA,
    and keep PC1 as a single representative column."""
    pc1_df = pd.DataFrame(index=df.index)

    for group_name, cols in groups.items():
        cols = [c for c in cols if c in df.columns]

        if len(cols) == 1:
            pc1_df[group_name] = StandardScaler().fit_transform(df[[cols[0]]])
            continue

        pipeline = make_pipeline(
            SimpleImputer(strategy="mean"),
            StandardScaler(),
            PCA(n_components=1, random_state=0),
        )
        pc1 = pipeline.fit_transform(df[cols])
        pc1_df[group_name] = pc1.ravel()


        corr = np.corrcoef(pc1_df[group_name], df[cols].mean(axis=1, skipna=True))[0, 1]
        if corr < 0:
            pc1_df[group_name] *= -1

    return pc1_df

combined_df
pca_df = pca_reduce_groups(combined_df, experiment_to_columns)
X = pca_df.to_numpy()
# standardised
X -= X.mean(axis=0)
X /= X.std(axis=0, ddof=0)
y = combined_df['AgeInMonths'].to_numpy()
y = y - np.mean(y)
```

Below is the proposed PSI pipeline

```python
from __future__ import annotations
import numpy as np
from dataclasses import dataclass
from typing import Optional, Sequence, Dict, Tuple, List
from sklearn.linear_model import Lasso
from sklearn.model_selection import KFold
from scipy.stats import truncnorm
from scipy.optimize import brentq
import warnings
import time


# --------------------------------------------------------------------------
# Plain LASSO fitting, CV, and the Lee-Sun-Sun-Taylor (2013) polyhedron
# --------------------------------------------------------------------------

def fit_lasso(X: np.ndarray, y: np.ndarray, lam: float, tol: float = 1e-9,
              max_iter: int = 20000) -> np.ndarray:
    """beta_hat = argmin_b (1/2)||y - Xb||_2^2 + lam*||b||_1 (Lee et al.'s
    convention). sklearn's Lasso uses (1/(2n))||y-Xb||^2 + alpha*||b||_1, so
    alpha = lam / n."""
    n = X.shape[0]
    if lam <= 0:
        # OLS fallback (degenerate, shouldn't really happen with our grids)
        beta, *_ = np.linalg.lstsq(X, y, rcond=None)
        return beta
    model = Lasso(alpha=lam / n, fit_intercept=False, tol=tol,
                   max_iter=max_iter)
    model.fit(X, y)
    return model.coef_


def active_set(beta: np.ndarray, tol: float = 1e-8) -> Tuple[np.ndarray, np.ndarray]:
    idx = np.where(np.abs(beta) > tol)[0]
    signs = np.sign(beta[idx]).astype(int)
    return idx, signs


def default_lambda_grid(X: np.ndarray, y: np.ndarray, n_grid: int = 15) -> np.ndarray:
    """Grid from just-below-lambda_max down to a small fraction of it."""
    lam_max = np.max(np.abs(X.T @ y))
    return lam_max * np.geomspace(1.0, 0.02, n_grid)


def cv_select_lambda_index(X: np.ndarray, y: np.ndarray, lambda_grid: np.ndarray,
                            K: int, rng: np.random.Generator) -> int:
    n = X.shape[0]
    kf = KFold(n_splits=K, shuffle=True, random_state=int(rng.integers(1 << 31)))
    splits = list(kf.split(X))
    scores = np.zeros(len(lambda_grid))
    for i, lam in enumerate(lambda_grid):
        errs = []
        for tr, te in splits:
            b = fit_lasso(X[tr], y[tr], lam)
            pred = X[te] @ b
            errs.append(np.mean((y[te] - pred) ** 2))
        scores[i] = np.mean(errs)
    return int(np.argmin(scores))


def lasso_polyhedron(X: np.ndarray, lam: float, E: np.ndarray,
                      signs: np.ndarray, tol_pinv: float = 1e-10
                      ) -> Tuple[np.ndarray, np.ndarray]:
    """Exact affine selection region {A y <= b} for the LASSO event
    supp(beta_hat(y,lam)) = E, sign(beta_hat_E) = signs (Lee, Sun, Sun &
    Taylor 2013, "Exact post-selection inference with the lasso").

    Uses the convention beta_hat = argmin (1/2)||y-Xb||^2 + lam||b||_1.
    """
    n, p = X.shape
    XE = X[:, E]
    rest = np.setdiff1d(np.arange(p), E)
    XEtXE_inv = np.linalg.pinv(XE.T @ XE, rcond=tol_pinv)
    SE = np.diag(signs.astype(float))

    # Active-set sign constraints: SE * XEtXE_inv * (XE^T y - lam*signs) >= 0
    #  <=>  -SE @ XEtXE_inv @ XE^T y  <=  -lam * SE @ XEtXE_inv @ signs
    A1 = -SE @ XEtXE_inv @ XE.T
    b1 = -lam * (SE @ XEtXE_inv @ signs)

    rows_A = [A1]
    rows_b = [b1]

    if rest.size > 0:
        Xr = X[:, rest]
        P_E = XE @ XEtXE_inv @ XE.T
        A0 = Xr.T @ (np.eye(n) - P_E)                       # (p-|E|) x n
        offset = lam * (Xr.T @ XE @ XEtXE_inv @ signs)       # (p-|E|,)
        # -lam <= A0 y + offset <= lam
        A_upper = A0
        b_upper = lam * np.ones(rest.size) - offset
        A_lower = -A0
        b_lower = lam * np.ones(rest.size) + offset
        rows_A += [A_upper, A_lower]
        rows_b += [b_upper, b_lower]

    A = np.vstack(rows_A)
    b = np.concatenate(rows_b)
    return A, b


def check_polyhedron(A: np.ndarray, b: np.ndarray, y: np.ndarray,
                      slack: float = 1e-6) -> bool:
    return bool(np.all(A @ y <= b + slack))


# --------------------------------------------------------------------------
# Coordinate-wise truncated-Gaussian Gibbs sampling within {A y <= b}
# --------------------------------------------------------------------------

def _coord_bounds(A_col: np.ndarray, resid: np.ndarray, b: np.ndarray
                   ) -> Tuple[float, float]:
    """For constraints A[:,i]*y_i <= b - A@y_{-i} (=: resid, precomputed as
    b - A@y with the i-th term's contribution removed by caller), return the
    feasible [lo, hi] interval for y_i."""
    lo, hi = -np.inf, np.inf
    pos = A_col > 1e-12
    neg = A_col < -1e-12
    if np.any(pos):
        hi = np.min(resid[pos] / A_col[pos])
    if np.any(neg):
        lo = np.max(resid[neg] / A_col[neg])
    return lo, hi


def truncated_normal_rvs(mean: float, sd: float, lo: float, hi: float,
                          rng: np.random.Generator) -> float:
    if hi <= lo:
        # numerically degenerate; snap to the (essentially single) feasible point
        return 0.5 * (lo + hi)
    a, bb = (lo - mean) / sd, (hi - mean) / sd
    return float(truncnorm.rvs(a, bb, loc=mean, scale=sd,
                                random_state=rng))


def gibbs_sweep_within_polyhedron(y: np.ndarray, mean: np.ndarray, sd: float,
                                   A: np.ndarray, b: np.ndarray,
                                   rng: np.random.Generator,
                                   n_sweeps: int = 1,
                                   coord_order: Optional[np.ndarray] = None
                                   ) -> np.ndarray:
    """One or more full coordinate sweeps of Gibbs sampling for
    y ~ N(mean, sd^2 I) truncated to {A y <= b}, updating y in place
    (and returning it)."""
    y = y.copy()
    n = y.shape[0]
    Ay = A @ y  # maintained incrementally
    idx_all = np.arange(n) if coord_order is None else coord_order
    for _ in range(n_sweeps):
        for i in idx_all:
            A_col = A[:, i]
            # resid = b - (A@y - A_col*y_i) = b - Ay + A_col*y_i
            resid = b - Ay + A_col * y[i]
            lo, hi = _coord_bounds(A_col, resid, b)
            new_yi = truncated_normal_rvs(mean[i], sd, lo, hi, rng)
            Ay += A_col * (new_yi - y[i])
            y[i] = new_yi
    return y


# --------------------------------------------------------------------------
# The rest of the two-stage randomization machinery
# --------------------------------------------------------------------------

@dataclass
class SelectionEvent:
    lambda_idx: int
    lam: float
    E: np.ndarray
    signs: np.ndarray
    A: np.ndarray
    b: np.ndarray


def run_selection(X, y, lambda_grid, sigma1, sigma2_cv, sigma2_select, K, rng
                   ) -> Tuple[SelectionEvent, np.ndarray, np.ndarray, np.ndarray]:
    n = X.shape[0]
    y_inter = y + rng.normal(scale=sigma1, size=n)
    y_cv = y_inter + rng.normal(scale=sigma2_cv, size=n)
    lam_idx = cv_select_lambda_index(X, y_cv, lambda_grid, K, rng)
    lam = lambda_grid[lam_idx]
    y_select = y_inter + rng.normal(scale=sigma2_select, size=n)
    beta = fit_lasso(X, y_select, lam)
    E, signs = active_set(beta)
    A, b = (None, None)
    if len(E) > 0:
        A, b = lasso_polyhedron(X, lam, E, signs)
    ev = SelectionEvent(lam_idx, lam, E, signs, A, b)
    return ev, y_inter, y_cv, y_select


# --------------------------------------------------------------------------
# The Gibbs sampler proper
# --------------------------------------------------------------------------

@dataclass
class VariableResult:
    variable: int
    beta_ols: float
    pvalue: float
    ci: Tuple[float, float]
    n_gibbs_samples: int
    umpu: bool


@dataclass
class GibbsSelectiveInferenceResult:
    selected: np.ndarray
    signs: np.ndarray
    lam: float
    per_variable: Dict[int, "VariableResult"]
    sigma_noise_hat: float
    condition_on_lambda: bool
    alpha: float = 0.10

    def summary(self) -> str:
        header = (f"{'var':>4} {'beta_ols':>10} {'p-value':>9} "
                   f"{'CI_low':>10} {'CI_high':>10}  n_samp")
        lines = [header, "-" * len(header)]
        for j, r in self.per_variable.items():
            lines.append(
                f"{j:>4} {r.beta_ols:>10.4f} {r.pvalue:>9.4f} "
                f"{r.ci[0]:>10.4f} {r.ci[1]:>10.4f}  {r.n_gibbs_samples}"
            )
        return "\n".join(lines)

    def print_latex(self, variable_names=None):
        """
        Print a LaTeX-ready summary table.
    
        Parameters
        ----------
        variable_names : list of str, optional
            Names corresponding to the variables in `self.selected`.
            If None, uses "Variable 0", "Variable 1", etc.
        """
        if variable_names is not None:
            if len(variable_names) != len(self.selected):
                raise ValueError(
                    "variable_names must have the same length as self.selected."
                )
            names = variable_names
        else:
            names = [f"Variable {j}" for j in self.selected]
    
        print(r"\begin{table}[ht]")
        print(r"    \centering")
        print(r"    \caption{Selective inference results}")
        print(r"    \label{tab:selective_inference}")
        print(r"    \begin{tabular}{lrrrr}")
        print(r"        \toprule")
        print(
            r"        Variable & $\hat{\beta}_{\mathrm{OLS}}$ & "
            r"$p$-value & CI Lower & CI Upper \\"
        )
        print(r"        \midrule")
    
        for name, j in zip(names, self.selected):
            result = self.per_variable[j]
    
            print(
                f"        {name} & "
                f"${result.beta_ols:.4f}$ & "
                f"${result.per_variable['pvalue']::.4f}$ & "
                f"${result.per_variable['ci'][0]:.4f}$ & "
                f"${result.per_variable['ci'][1]::.4f}$ \\\\"
            )
    
        print(r"        \bottomrule")
        print(r"    \end{tabular}")
        print(r"\end{table}")


class GibbsCVLassoSI:
    """
    Gibbs sampler for selective inference after CV-selected LASSO, following
    the four-step recipe at the end of Sec 6.1 of Tian & Taylor (2016).
    See the module docstring for the (LASSO vs sqrt-LASSO) deviation.

    Parameters
    ----------
    n_gibbs : total Gibbs sweeps to run (after burn-in) per tested variable.
    burn_in : sweeps discarded at the start of each chain.
    thin    : keep every `thin`-th sweep (reduces autocorrelation before we
              treat samples as ~iid for the empirical selective law).
    cv_update_every : refresh y_CV (the expensive rejection-sampled step)
              only every this-many sweeps, exactly as the paper recommends
              ("we do not update y_CV too often"). Ignored if
              condition_on_lambda=False.
    yselect_sweeps_per_iter : number of coordinate sweeps of the
              y_select | polyhedron step per outer Gibbs iteration (this
              step is cheap and closed-form, so a few sweeps per outer
              iteration improves mixing at negligible cost).
    condition_on_lambda : if True, also resample y_CV and require the CV-
              selected lambda index to match (faithful to the paper, but the
              y_CV step still requires expensive rejection sampling -- no
              way around that without an explicit description of the CV
              selection event, which the paper does not provide either).
              If False, lambda is fixed at its observed value (skips y_CV
              entirely) -- much faster, common approximation.
    max_cv_rejection_proposals : safety cap for the y_CV rejection step.
    use_umpu : if True, use the Lehmann-Scheffe/UMPU cutoffs (via importance
              reweighting of the Gibbs draws) instead of equal-tailed
              cutoffs for the p-value/CI.
    """

    def __init__(self,
                 lambda_grid: Optional[np.ndarray] = None,
                 sigma1: Optional[float] = None,
                 sigma2_cv: Optional[float] = None,
                 sigma2_select: Optional[float] = None,
                 noise_frac: float = 0.75,
                 K: int = 5,
                 condition_on_lambda: bool = False,
                 n_gibbs: int = 500,
                 burn_in: int = 100,
                 thin: int = 2,
                 cv_update_every: int = 10,
                 yselect_sweeps_per_iter: int = 1,
                 max_cv_rejection_proposals: int = 500,
                 alpha: float = 0.05,
                 use_umpu: bool = True,
                 seed: int = 0,
                 verbose: bool = False):
        self.lambda_grid = lambda_grid
        self.sigma1 = sigma1
        self.sigma2_cv = sigma2_cv
        self.sigma2_select = sigma2_select
        self.noise_frac = noise_frac
        self.K = K
        self.condition_on_lambda = condition_on_lambda
        self.n_gibbs = n_gibbs
        self.burn_in = burn_in
        self.thin = thin
        self.cv_update_every = cv_update_every
        self.yselect_sweeps_per_iter = yselect_sweeps_per_iter
        self.max_cv_rejection_proposals = max_cv_rejection_proposals
        self.alpha = alpha
        self.use_umpu = use_umpu
        self.seed = seed
        self.verbose = verbose

    # ---------------- the four Gibbs steps ----------------

    def _resample_yCV(self, X, yinter, obs_lambda_idx, lambda_grid, sigma2_cv,
                       K, rng) -> Optional[np.ndarray]:
        n = X.shape[0]
        for _ in range(self.max_cv_rejection_proposals):
            cand = yinter + rng.normal(scale=sigma2_cv, size=n)
            idx = cv_select_lambda_index(X, cand, lambda_grid, K, rng)
            if idx == obs_lambda_idx:
                return cand
        return None  # give up this round, keep old value

    def _run_chain(self, X, y_perp, c, eta, cc, m_ref, sd_t, sigma_noise_hat,
                   sigma1, sigma2_cv, sigma2_select, obs_ev: SelectionEvent,
                   y_select_init, y_cv_init, rng) -> np.ndarray:
        """Runs the 4-block Gibbs chain whose stationary law for t=c@y is
        the selective law f*_{m_ref}(t) (Lemma 2), i.e. as if H0j: beta_j =
        m_ref were true. We pick m_ref = t_obs for good mixing/acceptance,
        then use importance reweighting (exponential tilting, exact by
        Lemma 2's exponential-family structure) to evaluate the pivot at
        any other hypothesized mean (in particular 0, for the p-value) --
        see the UMPU helper methods below.
        
        RETURNS SORTED VERSION OF ARRAY"""
        n = X.shape[0]
        A, b = obs_ev.A, obs_ev.b
        lam = obs_ev.lam

        t = float(rng.normal(m_ref, sd_t))    # start near the reference mean
        y_full = y_perp + t * eta
        y_select = y_select_init.copy()
        y_cv = y_cv_init.copy() if y_cv_init is not None else y_full.copy()
        y_inter = y_full.copy()

        prec_y = 1.0 / sigma_noise_hat ** 2
        prec_1 = 1.0 / sigma1 ** 2
        prec_cv = 1.0 / sigma2_cv ** 2
        prec_sel = 1.0 / sigma2_select ** 2

        samples = []
        n_total = self.burn_in + self.n_gibbs * self.thin
        for it in range(n_total):
            # --- step: y_CV (expensive; refresh infrequently) ---
            if self.condition_on_lambda and (it % self.cv_update_every == 0):
                cand = self._resample_yCV(X, y_inter, obs_ev.lambda_idx,
                                           self.lambda_grid_, sigma2_cv, self.K, rng)
                if cand is not None:
                    y_cv = cand

            # --- step: y_select | polyhedron (E, s_E) ---
            y_select = gibbs_sweep_within_polyhedron(
                y_select, y_inter, sigma2_select, A, b, rng,
                n_sweeps=self.yselect_sweeps_per_iter)

            # --- step: y_inter | y, y_CV, y_select (closed form) ---
            denom = prec_1 + prec_cv + prec_sel
            mean_inter = (prec_1 * y_full + prec_cv * y_cv + prec_sel * y_select) / denom
            sd_inter = np.sqrt(1.0 / denom)
            y_inter = mean_inter + sd_inter * rng.normal(size=n)

            # --- step: y | y_inter (closed form, nuisance fixed) ---
            # Paper's full-vector formula: y|y_inter ~ N( (beta_j/sigma^2 +
            # y_inter/sigma1^2)/(1/sigma^2+1/sigma1^2), isotropic variance ),
            # projected onto eta=c (NOT the normalized c/cc direction) since
            # eta_j^T y is the sufficient statistic t := c @ y. Projecting an
            # isotropic N(mu, s^2 I) onto direction c gives
            # N(c@mu, s^2 * (c@c)) = N(c@mu, s^2 * cc).
            denom_y = prec_y + prec_1
            c_dot_inter = float(c @ y_inter)
            mean_t = (prec_y * m_ref + prec_1 * c_dot_inter) / denom_y
            sd_t_cond = np.sqrt(cc / denom_y)
            t = float(rng.normal(mean_t, sd_t_cond))
            y_full = y_perp + t * eta

            if it >= self.burn_in and (it - self.burn_in) % self.thin == 0:
                samples.append(t)

        return np.sort(np.array(samples))

    # ---------------- UMPU / importance-reweighting machinery ----------------

    @staticmethod
    def _iw(t_samples, m_ref, theta, sd_t):
        z = (theta - m_ref) * t_samples / (sd_t ** 2)
        z = z - np.max(z)
        return np.exp(z)

    @staticmethod
    def _wcdf(t_samples, w, x):
        tot = w.sum()
        return 0.5 if tot <= 0 else float(w[t_samples <= x].sum() / tot)

    @staticmethod
    def _wmean(t_samples, w):
        tot = w.sum()
        return 0.0 if tot <= 0 else float((w * t_samples).sum() / tot)

    @staticmethod
    def _wmean_restricted(t_samples, w, lo, hi):
        tot = w.sum()
        if tot <= 0:
            return 0.0
        mask = (t_samples >= lo) & (t_samples <= hi)
        return float((w[mask] * t_samples[mask]).sum() / tot)

    def _umpu_cutoffs(self, t_samples, m_ref, theta0, sd_t, alpha):
        """Solve the two Lehmann-Scheffe/UMPU equations using the
        importance-reweighted empirical law under theta0.
    
        Returns (c1, c2), or (None, None) if no empirical solution exists.
        """
        if len(t_samples) == 0:
            return None, None
    
        # ---------------------------------------------------------------
        # Compute importance weights ONCE
        # ---------------------------------------------------------------
        w = self._iw(t_samples, m_ref, theta0, sd_t)
    
        if not np.all(np.isfinite(w)):
            return None, None
    
        total_w = w.sum()
    
        if total_w <= 0:
            return None, None

        t = t_samples
    
    
        # ---------------------------------------------------------------
        # Cumulative quantities ONCE
        # ---------------------------------------------------------------
        cum_w = np.cumsum(w)
        cum_tw = np.cumsum(t * w)
    
        # Weighted mean E[T]
        Ef = cum_tw[-1] / total_w
    
        # ---------------------------------------------------------------
        # Weighted CDF
        # ---------------------------------------------------------------
        def cdf(x):
            # Number of observations <= x
            idx = np.searchsorted(t, x, side="left")
    
            if idx == 0:
                return 0.0
    
            return cum_w[idx - 1] / total_w
    
        # ---------------------------------------------------------------
        # Weighted quantile
        # ---------------------------------------------------------------
        def weighted_quantile(q):
            q = np.clip(q, 0.0, 1.0)
    
            idx = np.searchsorted(
                cum_w / total_w,
                q,
                side="left"
            )
    
            idx = min(idx, len(t) - 1)
    
            return float(t[idx])
    
        # ---------------------------------------------------------------
        # c1 must satisfy F(c1) <= alpha
        # ---------------------------------------------------------------
        c1_upper = weighted_quantile(alpha)
    
        # Candidate c1 values must actually satisfy F(c1) <= alpha.
        candidates = np.unique(t)
        
        candidates = candidates[
            np.array([cdf(x) <= alpha for x in candidates])
        ]
        
        if len(candidates) == 0:
            return None, None
    

        def restricted_mean(c1, c2):
            # Indices corresponding to
            # c1 <= T <= c2
            left = np.searchsorted(t, c1, side="left")
            right = np.searchsorted(t, c2, side="right")
    
            if left >= right:
                return 0.0
    
            total_tw = cum_tw[right - 1]
    
            if left > 0:
                total_tw -= cum_tw[left - 1]
    
            return total_tw / total_w
    

        target = (1.0 - alpha) * Ef
    
        best_residual = np.inf
        best_c1 = None
        best_c2 = None
    
        for c1 in candidates:
    
            # Size equation
            target_cdf = 1.0 - alpha + cdf(c1)
    
            if target_cdf > 1.0:
                c2 = np.inf
            elif target_cdf < 0.0:
                c2 = -np.inf
            else:
                c2 = weighted_quantile(target_cdf)
    
            if c2 is None:
                continue
    
            # Unbiasedness equation
            Er = restricted_mean(c1, c2)
    
            residual = Er - target
    
            if np.isfinite(residual) and abs(residual) < best_residual:
                best_residual = abs(residual)
                best_c1 = float(c1)
                best_c2 = float(c2)
    
        if best_c1 is None:
            return None, None
    
        return best_c1, best_c2

    def _pvalue_umpu(self, t_samples, m_ref, t_obs, sd_t,
                 theta0=0.0, min_alpha=None):

        n = len(t_samples)
        a_lo = min_alpha if min_alpha is not None else max(4.0 / n, 1e-3)
    
        # Candidate alpha values.
        alpha_grid = np.linspace(a_lo, 1.0 - a_lo, 200)
    
        for alpha in alpha_grid:
            c1, c2 = self._umpu_cutoffs(
                t_samples, m_ref, theta0, sd_t, alpha
            )
    
            if c1 is None or c2 is None:
                continue
    
            if t_obs < c1 or t_obs > c2:
                return float(alpha)
    
        return 1.0

    def _ci_umpu(self, t_samples, m_ref, t_obs, sd_t, alpha):

        width = 6 * sd_t
        n_grid = 10001
    
        grid = np.linspace(t_obs - width, t_obs + width, n_grid)
    
        # Find the grid point closest to t_obs
        center = np.searchsorted(grid, t_obs)
        
        # Make sure t_obs itself is accepted
        c1, c2 = self._umpu_cutoffs(
            t_samples, m_ref, t_obs, sd_t, alpha
        )
    
        if c1 is None or c2 is None or not (c1 <= t_obs <= c2):
            return -np.inf, np.inf
    
        # Search left
        lo_idx = center
        for k in range(center - 1, -1, -1):
            c1, c2 = self._umpu_cutoffs(
                t_samples, m_ref, grid[k], sd_t, alpha
            )
            if c1 is None or c2 is None or not (c1 <= t_obs <= c2):
                break
            lo_idx = k
    
        # Search right
        hi_idx = center
        for k in range(center + 1, n_grid):
            c1, c2 = self._umpu_cutoffs(
                t_samples, m_ref, grid[k], sd_t, alpha
            )
            if c1 is None or c2 is None or not (c1 <= t_obs <= c2):
                break
            hi_idx = k
    
        if lo_idx == 0:
            lo = -np.inf
        else:
            lo = grid[lo_idx]
    
        if hi_idx == n_grid - 1:
            hi = np.inf
        else:
            hi = grid[hi_idx]
    
        return lo, hi

    @staticmethod
    def _robust_root(f, a, b, tries=3):
        fa, fb = f(a), f(b)
        k = 0
        while fa * fb > 0 and k < tries:
            a -= (b - a) / 2
            b += (b - a) / 2
            fa, fb = f(a), f(b)
            k += 1
        if fa * fb > 0:
            return (a + b) / 2
        try:
            return brentq(f, a, b, xtol=1e-3)
        except ValueError:
            return (a + b) / 2

    # ---------------- fit ----------------

    def fit(self, X: np.ndarray, y: np.ndarray) -> GibbsSelectiveInferenceResult:
        t_start = time.time()
        X = np.asarray(X, dtype=float)
        y = np.asarray(y, dtype=float)
        n, p = X.shape
        rng = np.random.default_rng(self.seed)

        if n > p:
            beta_full = np.linalg.lstsq(X, y, rcond=None)[0]
            resid = y - X @ beta_full
            sigma_noise_hat = np.std(resid, ddof=p) if n - p > 0 else np.std(resid)
        else:
            sigma_noise_hat = np.std(y)
        sigma_noise_hat = max(sigma_noise_hat, 1e-6)

        sigma1 = self.sigma1 or self.noise_frac * sigma_noise_hat
        sigma2_cv = self.sigma2_cv or self.noise_frac * sigma_noise_hat
        sigma2_select = self.sigma2_select or self.noise_frac * sigma_noise_hat
        self.lambda_grid_ = (self.lambda_grid if self.lambda_grid is not None
                              else default_lambda_grid(X, y))

        obs_ev, y_inter_obs, y_cv_obs, y_select_obs = run_selection(
            X, y, self.lambda_grid_, sigma1, sigma2_cv, sigma2_select, self.K, rng)

        if len(obs_ev.E) == 0:
            warnings.warn("No variables selected; returning empty result.")
            return GibbsSelectiveInferenceResult(
                obs_ev.E, obs_ev.signs, obs_ev.lam, {}, sigma_noise_hat,
                self.condition_on_lambda, self.alpha)

        E, signs, lam = obs_ev.E, obs_ev.signs, obs_ev.lam
        if self.verbose:
            print(f"[fit] selected {len(E)} vars at lambda={lam:.4g} "
                  f"({time.time()-t_start:.1f}s so far)")

        XE = X[:, E]
        pinvXE = np.linalg.pinv(XE)
        beta_ols = pinvXE @ y

        print(f'Selected Set Indeces: {E}')
        per_variable: Dict[int, VariableResult] = {}
        for jpos, j in enumerate(E):
            c = pinvXE[jpos]
            cc = float(c @ c)
            t_obs = float(c @ y)
            y_perp = y - (t_obs / cc) * c
            eta = c / cc
            sd_t = sigma_noise_hat * np.sqrt(cc)

            t0 = time.time()
            samples_ref = self._run_chain(
                X, y_perp, c, eta, cc, t_obs, sd_t, sigma_noise_hat,
                sigma1, sigma2_cv, sigma2_select, obs_ev,
                y_select_obs, y_cv_obs, rng)
            samples_null = self._run_chain(
                X, y_perp, c, eta, cc, 0.0, sd_t, sigma_noise_hat,
                sigma1, sigma2_cv, sigma2_select, obs_ev,
                y_select_obs, y_cv_obs, rng)
            if self.verbose:
                print(f"[fit] var {j}: {len(samples_ref)}+{len(samples_null)} "
                      f"Gibbs samples in {time.time()-t0:.1f}s "
                      f"(ref mean {samples_ref.mean():.3f}, "
                      f"null-chain mean {samples_null.mean():.3f}, "
                      f"sd_t={sd_t:.3f})")

            if self.use_umpu:
                t1 = time.time()
                pvalue = self._pvalue_umpu(samples_null, 0.0, t_obs, sd_t)
                print(f'UMPU p-value successfully computed: {pvalue}. It took {time.time() - t1} seconds')
                t1 = time.time()
                ci = self._ci_umpu(samples_ref, t_obs, t_obs, sd_t, self.alpha)
                print(f'UMPU CI successfully computed: {ci}. It took {time.time() - t1} seconds')
            else:
                w0 = self._iw(samples_null, 0.0, 0.0, sd_t)  # trivial (weights=1)
                F0 = self._wcdf(samples_null, w0, t_obs)
                pvalue = 2 * min(F0, 1 - F0)

                def f_lo(m0):
                    w = self._iw(samples_ref, t_obs, m0, sd_t)
                    return self._wcdf(samples_ref, w, t_obs) - (1 - self.alpha / 2)

                def f_hi(m0):
                    w = self._iw(samples_ref, t_obs, m0, sd_t)
                    return self._wcdf(samples_ref, w, t_obs) - self.alpha / 2
                width = 8 * sd_t
                lo = self._robust_root(f_lo, t_obs - width, t_obs + width)
                hi = self._robust_root(f_hi, t_obs - width, t_obs + width)
                ci = tuple(sorted([lo, hi]))

            per_variable[j] = VariableResult(
                variable=j, beta_ols=float(beta_ols[jpos]), pvalue=pvalue,
                ci=ci, n_gibbs_samples=len(samples_ref), umpu=self.use_umpu)
            print(f'Finished Processing Variable: {j}, beta_OLS = {float(beta_ols[jpos])}')

        result = GibbsSelectiveInferenceResult(
            E, signs, lam, per_variable, sigma_noise_hat,
            self.condition_on_lambda, self.alpha)
        self.result_ = result
        if self.verbose:
            print(f"[fit] total time {time.time()-t_start:.1f}s")
        return result
```

Run the pipeline
